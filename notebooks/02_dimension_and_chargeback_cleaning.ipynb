```python
raw_txn_ids = set(transactions_raw["txn_id"])

orphan_in_raw = orphan_chargebacks[
    orphan_chargebacks["txn_id"].isin(raw_txn_ids)
].copy()

print("=" * 60)
print("ORPHAN CHARGEBACK TRACE")
print("=" * 60)

print("Total orphan chargebacks:", len(orphan_chargebacks))
print("Found in raw transactions:", len(orphan_in_raw))
print("Not found in raw transactions:",
      len(orphan_chargebacks) - len(orphan_in_raw))
```

```python
if len(orphan_in_raw) > 0:
    orphan_raw_details = transactions_raw[
        transactions_raw["txn_id"].isin(orphan_in_raw["txn_id"])
    ].copy()

    display(orphan_raw_details)
else:
    print("None of the orphan chargebacks exist in the raw transaction file.")
```

```python
# Identify the cleaning-rule reason for each orphan transaction

orphan_txns = transactions_raw[
    transactions_raw["txn_id"].isin(orphan_chargebacks["txn_id"])
].copy()

orphan_txns["invalid_amount"] = orphan_txns["amount"] <= 0
orphan_txns["invalid_merchant"] = ~orphan_txns["merchant_id"].isin(
    merchants["merchant_id"]
)

orphan_txns["invalid_card"] = ~orphan_txns["card_id"].isin(
    cards["card_id"]
)

orphan_txns["invalid_customer"] = ~orphan_txns["customer_id"].isin(
    customers["customer_id"]
)

orphan_txns["invalid_device"] = (
    ~orphan_txns["device_id"].fillna(-1).isin(devices_clean["device_id"])
)

display(
    orphan_txns[
        [
            "txn_id",
            "amount",
            "merchant_id",
            "invalid_amount",
            "invalid_merchant",
            "invalid_card",
            "invalid_customer",
            "invalid_device"
        ]
    ]
)
```

```python
print("Cleaning reasons among orphan chargebacks:")
print()

print(
    "Invalid amount:",
    orphan_txns["invalid_amount"].sum()
)

print(
    "Invalid merchant:",
    orphan_txns["invalid_merchant"].sum()
)

print(
    "Invalid card:",
    orphan_txns["invalid_card"].sum()
)

print(
    "Invalid customer:",
    orphan_txns["invalid_customer"].sum()
)

print(
    "Invalid device:",
    orphan_txns["invalid_device"].sum()
)
```

```python
orphan_financial_impact = orphan_chargebacks[
    ["txn_id", "loss_amount", "recovered_amount"]
].copy()

orphan_financial_impact["net_loss"] = (
    orphan_financial_impact["loss_amount"]
    - orphan_financial_impact["recovered_amount"]
)

print("=" * 60)
print("ORPHAN CHARGEBACK FINANCIAL IMPACT")
print("=" * 60)

print("Total loss:", orphan_financial_impact["loss_amount"].sum())
print("Total recovered:", orphan_financial_impact["recovered_amount"].sum())
print("Net loss:", orphan_financial_impact["net_loss"].sum())

display(orphan_financial_impact)
```
